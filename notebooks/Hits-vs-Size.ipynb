```python
# import necessary packages
import requests
```

```python
# Official NHL API URL BASE
URL = "https://api-web.nhle.com/v1/"
```

```python
# Pull every single active player from the NHL API
try:

    # Make a GET request to the NHL API to retrieve player data
    # https://search.d3.nhle.com/api/v1/search/player?culture=en-us&limit=50000&q=*&active=true
    response = requests.get("https://search.d3.nhle.com/api/v1/search/player?culture=en-us&limit=50000&q=*&active=true")

    # Check if the request was successful (status code 200)
    print(response.status_code)

# Error handling for the request
except requests.exceptions.RequestException as e:
    print(f"Error: {e}")
```

```python
all_player_ids = response.json()
full_player_data = []
count = 0
len(all_player_ids)

```

```python
# for every single player, take their `playerId` and make a GET request to the NHL API to retrieve their player data
for player in all_player_ids:
    try:
        # make a get request to the NHL API to retrieve player data
        response = requests.get(f"{URL}player/{player['playerId']}/landing")

        # check if the request was successful (status code 200)

        if response.status_code == 200:
            count += 1
            print(count)
            full_player_data.append(response.json())

    except requests.exceptions.RequestException as e:
        print(f"Error: {e}")
```
